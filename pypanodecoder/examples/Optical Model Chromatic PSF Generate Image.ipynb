## Calculate sample PSF image on the image plane

Use the simple optical model to calculate and plot a sample PSF image, assuming a Cherenkov spectrum. The optics are described by the standard optical model data pack, from `resources/optical_model_data_pack.json`, which describes 

- the parameters of the ray tracing model, 
- the quantum efficency curve of the SiPM detetector, 
- the transmission curve of the lens,
- the atmospheric transmissivoty from the top of the atmosphere, and
- the refractive index of the lens as a function of photon energy.

The notebook loads the data pack and uses the thin-lens optical model to calculate and display the PSF image for off-axis rays (directed at pixel 4.5,5.5 in the camera).

```python
%matplotlib inline
import sys
sys.path.append("../..")

import numpy as np
import matplotlib.pyplot as plt

from pypanodecoder import optical_model
from pypanodecoder import plotting
```

```python
# Load the data pack which gives a dictionary of the items listed above
data_pack = optical_model.load_datapack()
```

```python
# Calculate the PSF images for rays whose principal rays are directed towards a location at 
# coordinates 4.5 and 5.5 (pixels) in the image plane, using 1 million rays.
nray,image,_,_,_,_ = optical_model.generate_psf_image(4.5,5.5,1000000,data_pack)
```

```python
plotting.plot_image(image/nray, plate_scale=1.0)
```
